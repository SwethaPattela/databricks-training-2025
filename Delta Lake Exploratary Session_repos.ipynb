```sql
-- Create target table
CREATE OR REPLACE TABLE my_training_schema.target_table (
  id INT,
  name STRING,
  status STRING
);
```

```sql

INSERT INTO my_training_schema.target_table VALUES
  (1, 'Alice', 'active'),
  (2, 'Bob', 'inactive'),
  (3, 'Charlie', 'active');
```

```sql
desc history my_training_schema.target_table
```

```sql
-- Create source table
CREATE OR REPLACE TABLE my_training_schema.source_table (
  id INT,
  name STRING,
  status STRING
);
```

```sql
INSERT INTO my_training_schema.source_table VALUES
  (2, 'Bob', 'active'),        -- Should update status
  (3, 'Charlie', 'deleted'),   -- Should delete
  (4, 'Diana', 'active');      -- Should insert
```

```sql
select * from my_training_schema.target_table
```

```sql


-- Merge statement demonstrating insert, update, and delete
MERGE INTO my_training_schema.target_table AS target
USING my_training_schema.source_table AS source
ON target.id = source.id
WHEN MATCHED AND source.status = 'deleted' THEN DELETE
WHEN MATCHED AND target.status <> source.status THEN
  UPDATE SET target.status = source.status
WHEN NOT MATCHED THEN
  INSERT (id, name, status) VALUES (source.id, source.name, source.status);

  --(2, 'Bob', 'active'),        -- Should update status
  --(3, 'Charlie', 'deleted'),   -- Should delete
 -- (4, 'Diana', 'active');  
```
